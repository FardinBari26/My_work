```python
sns.set_theme(style='whitegrid', context='notebook')
plt.rcParams['figure.figsize'] = (7, 4)
plt.rcParams['figure.dpi']     = 100
```

```python
gym_file = "/content/drive/MyDrive/Data_set/Class 5/gym_membership.csv"

gym_df = pd.read_csv(gym_file)
gym_df.head()
```

```python
# Bar chart of membership types
membership_count = gym_df['abonoment_type'].value_counts()
print(membership_count)

fig, ax = plt.subplots()
ax.bar(membership_count.index, membership_count.values, color=['#e91e63', '#2196f3'])
ax.set(title='Membership Types', ylabel='Count')
```

```python
# Histogram of avg_time_in_gym

fig, ax = plt.subplots()
ax.hist(gym_df['avg_time_in_gym'], bins=10, color='#2196f3')
ax.set(title='Distribution of Avg_time_in_gym',
       xlabel='Avg_time_in_gym', ylabel='Frequency')
plt.show()
```

```python
# Scatter plot of Age vs visit_per_week

fig, ax = plt.subplots()
ax.scatter(gym_df['Age'], gym_df['visit_per_week'], color='#2196f3')
ax.set(title='Age vs Visit_per_week',
       xlabel='Age', ylabel='Visit_per_week')
plt.show()
```

```python
# Box plot — gender vs Age

fig, ax = plt.subplots()

sns.boxplot(data=gym_df, x='gender', y='Age', ax=ax)
ax.set_title('gender vs Age')
plt.show()
```

```python
# Count plot — days_per_week

fig, ax = plt.subplots()

sns.countplot(data=gym_df, x='days_per_week', ax=ax)
ax.set_title('Days per week')
plt.show()
```

```python
# Bar plot — mean visit_per_week per abonoment_type

fig, ax = plt.subplots()

sns.barplot(data=gym_df, x='abonoment_type', y='visit_per_week', ax=ax)
ax.set_title('Mean visit_per_week per abonoment_type')
```
